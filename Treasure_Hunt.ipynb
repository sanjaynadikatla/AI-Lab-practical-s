```python
# The map: each room lists the rooms you can walk to next.

map_rooms = {
    "A": ["B", "C"],   # from A you can go to B or C
    "B": ["D"],        # from B you can go to D
    "C": ["D"],        # from C you can go to D
    "D": ["G"],        # from D you can go to G
    "G": []            # G is the treasure! No need to go further.
}

print("Our map:")
print("          A (Start)")
print("         / \\")
print("        B   C")
print("         \\ /")
print("          D")
print("          |")
print("          G (Treasure 💎)")
print()

print("Doors from each room:", map_rooms)
```

```python


def bfs(start, goal):
    to_do = [start]      # rooms waiting to be checked
    visited = []         # rooms we have already checked
    order = []           # the order we checked rooms

    while to_do:
        # Take from the FRONT of the list
        room = to_do.pop(0)

        if room in visited:
            continue

        visited.append(room)
        order.append(room)

        if room == goal:
            return order      # found the treasure!

        for nxt in map_rooms[room]:
            to_do.append(nxt)

    return order


order = bfs("A", "G")
print("BFS checked the rooms in this order:", order)
print("Number of rooms BFS checked:", len(order))
```

```python
def dfs (start,goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop()
        if room in visited:
            continue
        visited.append(room)
        order.append(room)
        if room == goal:
            return order
        for neighbour in map_rooms[room]:
            to_do.append(neighbour)
    return order
order = dfs("A","G")
print("DFS order:", order)
```

```python
door_cost={("A","B"):1,
           ("B","D"):1,
           ("D","G"):1,
           ("A","C"):5,
           ("C","D"):1
          }
def path_cost(path):
    total=0
    for i in range(len(path) - 1):
        door =(path[i],path[i+1])
        total+= door_cost[door]
    return total
path1=["A","B","D","G"]
path2=["A","C","D","G"]
path1_cost=path_cost(path1)
path2_cost=path_cost(path2)
print("cost of path1: " ,path1_cost)
print("cost of path2: " ,path2_cost)
if path1_cost <= path2_cost :
    print("best path: ", path1 , "best cost: ",path1_cost)
else:
     print("best path: ", path2 , "best cost: ",path2_cost)
```
